```python
# Feature selection
df.columns
```

```python
x=df.iloc[:,:-1]
x
```

```python
x.columns
```

```python
y=df['Outcome']
y
```

```python
# train test split
from sklearn.model_selection import train_test_split
xtrain,xtest,ytrain,ytest=train_test_split(x,y,test_size=0.2,random_state=1)
```

gini index, entropy - quality of split

gini index - default

```python
from sklearn.tree import DecisionTreeClassifier
#destreclas=DecisionTreeClassifier(max_depth=4) # gini index
destreclas=DecisionTreeClassifier(criterion='gini',max_depth=3)
destreclas=destreclas.fit(xtrain,ytrain)
ypred=destreclas.predict(xtest)
```

```python
ypred # outcome
```

```python
from sklearn.metrics import accuracy_score,confusion_matrix
```

```python
print("Accuracy ",accuracy_score(ytest,ypred))
```

```python
cm=confusion_matrix(ytest,ypred)
print('Confusion Matrix')
print(cm)
```

```python
from sklearn import tree
```

```python
text_representation = tree.export_text(destreclas)
print(text_representation)
```

0 Pregnancies , 1 Glucose , 2 BloodPressure,
3 SkinThickness, 4 Insulin,
       5 BMI, 6 DiabetesPedigreeFunction , 7 Age

Target Class 0 - absence 1 presence

File - create - write

```python
with open('description_tree.log','w') as fout:
  fout.write(text_representation)
```

Tree visualization

```python
feature_names=['Pregnancies', 'Glucose', 'BloodPressure', 'SkinThickness', 'Insulin','BMI', 'DiabetesPedigreeFunction', 'Age']
target_names=['0','1']
fig=plt.figure(figsize=(23,30))
plot=tree.plot_tree(destreclas,feature_names=feature_names,class_names=target_names,filled=True)
fig.savefig('Tree_gini.png')
```

Entropy

```python
destreclas=DecisionTreeClassifier(criterion='entropy',max_depth=3)
destreclas=destreclas.fit(xtrain,ytrain)
ypred=destreclas.predict(xtest)
```

```python
print(accuracy_score(ytest,ypred))
```

Entropy gives good results compared to gini index
